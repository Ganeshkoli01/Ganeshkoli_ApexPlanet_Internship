```python
import pandas as pd
import numpy as np
import os

# Verify that the dataset is available
if "df" not in globals():
    raise NameError(
        "Please load your dataset into a DataFrame named df first."
    )

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
display(df.head())
```

```python
# Step 2: Missing values report

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isna().sum().values,
    "Missing_Percentage": (
        df.isna().mean().values * 100
    ).round(2),
    "Non_Missing_Count": df.notna().sum().values
})

missing_report = missing_report.sort_values(
    "Missing_Count",
    ascending=False
)

display(missing_report)
```

```python
# Step 3: Find exact duplicate rows

duplicate_count = df.duplicated().sum()

print("Total duplicate rows:", duplicate_count)

duplicate_rows = df[df.duplicated(keep=False)].copy()

print("Rows involved in duplicates:", len(duplicate_rows))
display(duplicate_rows.head(10))
```

```python
# Step 4: Check duplicate IDs if the column exists

if "Order_ID" in df.columns:
    duplicate_id_mask = df["Order_ID"].duplicated(keep=False)

    print("Rows with repeated Order_ID values:",
          duplicate_id_mask.sum())

    display(
        df.loc[duplicate_id_mask]
          .sort_values("Order_ID")
          .head(20)
    )
else:
    print("Order_ID column not found.")
```

```python
# Step 5: Inspect data types

dtype_report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Unique_Values": df.nunique(dropna=True).values
})

display(dtype_report)
```

```python
# Step 6: Check numerical ranges

numeric_checks = {
    "Age": lambda s: (s < 0) | (s > 120),
    "Quantity": lambda s: s <= 0,
    "Unit_Price": lambda s: s < 0,
    "Total_Sales": lambda s: s < 0
}

for column, rule in numeric_checks.items():
    if column in df.columns:
        values = pd.to_numeric(df[column], errors="coerce")
        invalid_mask = rule(values) & values.notna()

        print(f"\n{column} - suspicious values: {invalid_mask.sum()}")
        display(df.loc[invalid_mask].head(10))
```

```python
# Step 7: Inspect categorical columns

categorical_columns = [
    "Gender", "City", "Product", "Category"
]

for column in categorical_columns:
    if column in df.columns:
        print(f"\n--- {column} ---")
        print("Unique values:", df[column].nunique(dropna=True))
        display(
            df[column]
              .value_counts(dropna=False)
              .head(20)
        )
```

```python
# Step 8: Validate sales calculations

required_columns = ["Quantity", "Unit_Price", "Total_Sales"]

if all(column in df.columns for column in required_columns):
    quantity = pd.to_numeric(df["Quantity"], errors="coerce")
    unit_price = pd.to_numeric(df["Unit_Price"], errors="coerce")
    total_sales = pd.to_numeric(df["Total_Sales"], errors="coerce")

    expected_sales = quantity * unit_price

    valid_rows = (
        quantity.notna()
        & unit_price.notna()
        & total_sales.notna()
    )

    mismatch = (
        (total_sales - expected_sales).abs() > 0.02
    ) & valid_rows

    print("Rows checked:", valid_rows.sum())
    print("Sales calculation mismatches:", mismatch.sum())

    display(
        df.loc[mismatch, required_columns].head(10)
    )
else:
    print("One or more sales columns are missing.")
```

```python
# Step 9: Create the data quality summary

quality_checks = []

for column in df.columns:
    missing_count = int(df[column].isna().sum())

    quality_checks.append({
        "Check": f"Missing values - {column}",
        "Issue_Count": missing_count,
        "Status": "Review required" if missing_count else "Pass"
    })

quality_checks.extend([
    {
        "Check": "Exact duplicate rows",
        "Issue_Count": int(df.duplicated().sum()),
        "Status": "Review required" if df.duplicated().sum() else "Pass"
    }
])

if "Order_ID" in df.columns:
    duplicate_ids = int(
        df["Order_ID"].duplicated(keep=False).sum()
    )

    quality_checks.append({
        "Check": "Rows with repeated Order_ID",
        "Issue_Count": duplicate_ids,
        "Status": "Review required" if duplicate_ids else "Pass"
    })

data_quality_report = pd.DataFrame(quality_checks)

display(data_quality_report)
```

```python
# Step 10: Save Day 3 reports

os.makedirs("../reports", exist_ok=True)

missing_report.to_csv(
    "../reports/missing_values_report.csv",
    index=False
)

duplicate_rows.to_csv(
    "../reports/duplicate_rows.csv",
    index=False
)

data_quality_report.to_csv(
    "../reports/data_quality_report.csv",
    index=False
)

print("Day 3 reports saved successfully!")
print("Reports folder:", os.path.abspath("../reports"))
```

```python
#DAY 4
import pandas as pd
import numpy as np
import os

if "df" not in globals():
    raise NameError(
        "Load your original dataset into df before continuing."
    )

# Preserve the original dataset
df_original = df.copy(deep=True)

# Perform cleaning on a separate working copy
df_clean = df_original.copy(deep=True)

print("Original rows:", len(df_original))
print("Working-copy rows:", len(df_clean))
print("Original columns:", df_original.columns.tolist())
```

```python

# Count missing values before cleaning

missing_before = pd.DataFrame({
    "Column": df_clean.columns,
    "Missing_Before": df_clean.isna().sum().values,
    "Missing_Percentage": (
        df_clean.isna().mean().values * 100
    ).round(2)
})

display(
    missing_before.sort_values(
        "Missing_Before",
        ascending=False
    )
)
```

```python

# Fill missing Age values using the median

if "Age" in df_clean.columns:
    age_numeric = pd.to_numeric(
        df_clean["Age"], errors="coerce"
    )

    # Use the median of valid, plausible ages
    valid_age = age_numeric[
        age_numeric.between(0, 120)
    ]

    if age_numeric.isna().any() and not valid_age.empty:
        age_median = valid_age.median()

        df_clean["Age"] = age_numeric.fillna(age_median)

        print("Median age used:", age_median)
        print(
            "Missing Age after treatment:",
            df_clean["Age"].isna().sum()
        )
    else:
        print("No missing Age values to fill, or no valid ages.")
else:
    print("Age column not found.")
```

```python

# Fill missing City values with an explicit label

if "City" in df_clean.columns:
    missing_city_before = df_clean["City"].isna().sum()

    df_clean["City"] = df_clean["City"].fillna("Unknown")

    print("Missing City values before:", missing_city_before)
    print("Missing City values after:", df_clean["City"].isna().sum())
    print("\nCity value counts:")
    print(df_clean["City"].value_counts(dropna=False).head(15))
else:
    print("City column not found.")
```

```python

# Count exact duplicates before removal

duplicates_before = int(df_clean.duplicated().sum())

print("Exact duplicate rows:", duplicates_before)

# Keep the first occurrence of each exact duplicate row
df_clean = df_clean.drop_duplicates().copy()

duplicates_after = int(df_clean.duplicated().sum())

print("Duplicate rows after removal:", duplicates_after)
print("Rows remaining:", len(df_clean))
```

```python

# Review remaining missing values

remaining_missing = df_clean.isna().sum()

print("Remaining missing values by column:")
display(
    remaining_missing[
        remaining_missing > 0
    ].rename("Missing_Count").to_frame()
)
```

```python

# Create a cleaning log

cleaning_log = pd.DataFrame([
    {
        "Issue": "Missing Age",
        "Action": "Fill missing values with median of plausible ages, if applicable",
        "Reason": "Provides a robust numerical estimate",
        "Review_Required": "Yes - verify assumptions"
    },
    {
        "Issue": "Missing City",
        "Action": "Replace missing values with Unknown, if applicable",
        "Reason": "Preserves rows without inventing a city",
        "Review_Required": "Yes - confirm reporting requirements"
    },
    {
        "Issue": "Exact duplicate rows",
        "Action": "Keep first occurrence of exact duplicate rows",
        "Reason": "Avoids counting identical records twice when confirmed accidental",
        "Review_Required": "Yes - verify duplicates are accidental"
    }
])

display(cleaning_log)
```

```python

# Compare before and after cleaning

comparison = pd.DataFrame({
    "Metric": [
        "Total rows",
        "Total columns",
        "Exact duplicate rows",
        "Total missing cells"
    ],
    "Before_Cleaning": [
        len(df_original),
        len(df_original.columns),
        int(df_original.duplicated().sum()),
        int(df_original.isna().sum().sum())
    ],
    "After_Cleaning": [
        len(df_clean),
        len(df_clean.columns),
        int(df_clean.duplicated().sum()),
        int(df_clean.isna().sum().sum())
    ]
})

display(comparison)
```

```python

# Create output folders

os.makedirs("../data/processed", exist_ok=True)
os.makedirs("../reports", exist_ok=True)

# Save cleaned dataset
cleaned_path = "../data/processed/cleaned_sales_dataset.csv"
df_clean.to_csv(cleaned_path, index=False)

# Save cleaning log
log_path = "../reports/cleaning_log.csv"
cleaning_log.to_csv(log_path, index=False)

# Save a summary of changes
summary_path = "../reports/day4_cleaning_summary.txt"

summary = f"""
DAY 4 - DATA CLEANING SUMMARY

Original rows: {len(df_original)}
Cleaned rows: {len(df_clean)}
Rows removed: {len(df_original) - len(df_clean)}

Original exact duplicates: {df_original.duplicated().sum()}
Remaining exact duplicates: {df_clean.duplicated().sum()}

Missing cells before cleaning: {df_original.isna().sum().sum()}
Missing cells after cleaning: {df_clean.isna().sum().sum()}

Original columns: {len(df_original.columns)}
Cleaned columns: {len(df_clean.columns)}

Cleaning actions:
- Reviewed missing values.
- Applied the selected Age and City treatments when applicable.
- Removed exact duplicate rows from the working copy.
- Preserved the original DataFrame.
"""

with open(summary_path, "w", encoding="utf-8") as f:
    f.write(summary.strip())

print("Day 4 deliverables saved!")
print("Cleaned data:", os.path.abspath(cleaned_path))
print("Cleaning log:", os.path.abspath(log_path))
print("Summary:", os.path.abspath(summary_path))
```

```python

# DAY 5 - STEP 1: Create a safe working copy

df_day5 = df_clean.copy(deep=True)

print("Day 5 cleaning started")
print("Original dataset shape:", df.shape)
print("Day 4 cleaned dataset shape:", df_clean.shape)
print("Day 5 working dataset shape:", df_day5.shape)

assert df_day5.shape == df_clean.shape

display(df_day5.head())
```

```python

# DAY 5 - STEP 2: Standardize text

text_columns = [
    "Order_ID",
    "Customer_ID",
    "Customer_Name",
    "Gender",
    "City",
    "Product",
    "Category"
]

for column in text_columns:
    if column in df_day5.columns:
        df_day5[column] = (
            df_day5[column]
            .astype("string")
            .str.strip()
        )

# Standardize categorical labels
category_columns = ["Gender", "City", "Product", "Category"]

for column in category_columns:
    if column in df_day5.columns:
        df_day5[column] = (
            df_day5[column]
            .str.title()
        )

print("Text standardization completed.")

for column in category_columns:
    print(f"\n{column}:")
    print(sorted(
        df_day5[column].dropna().unique().tolist()
    ))
```

```python

# DAY 5 - STEP 3: Date conversion

original_dates = df_day5["Order_Date"].copy()

parsed_dates = pd.to_datetime(
    original_dates,
    errors="coerce"
)

# Identify dates that could not be parsed
invalid_date_mask = (
    original_dates.notna() & parsed_dates.isna()
)

invalid_dates = df_day5.loc[
    invalid_date_mask,
    ["Order_ID", "Order_Date"]
].copy()

print("Invalid dates found:", len(invalid_dates))

display(invalid_dates.head(20))

# Apply parsed dates
df_day5["Order_Date"] = parsed_dates

print("Order_Date data type:",
      df_day5["Order_Date"].dtype)

print("Earliest order:",
      df_day5["Order_Date"].min())

print("Latest order:",
      df_day5["Order_Date"].max())
```

```python

# DAY 5 - STEP 4: Numeric validation

numeric_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]

numeric_conversion_report = []

for column in numeric_columns:
    original = df_day5[column].copy()

    converted = pd.to_numeric(
        original,
        errors="coerce"
    )

    invalid_mask = original.notna() & converted.isna()

    numeric_conversion_report.append({
        "Column": column,
        "Invalid_Numeric_Values": int(invalid_mask.sum())
    })

    df_day5[column] = converted

numeric_report = pd.DataFrame(numeric_conversion_report)

display(numeric_report)

print("Numeric data types:")
print(df_day5[numeric_columns].dtypes)
```

```python

# DAY 5 - STEP 5: Investigate repeated order IDs

repeated_id_mask = df_day5.duplicated(
    subset=["Order_ID"],
    keep=False
)

repeated_order_rows = (
    df_day5.loc[repeated_id_mask]
    .sort_values("Order_ID")
    .copy()
)

print("Rows with repeated Order_ID:",
      len(repeated_order_rows))

print("Number of repeated Order_ID groups:",
      repeated_order_rows["Order_ID"].nunique())

display(repeated_order_rows)
```

```python

# DAY 5 - STEP 5A: Investigate repeated Order_ID

order_id_summary = (
    df_day5.groupby("Order_ID")
    .agg(
        Row_Count=("Order_ID", "size"),
        Unique_Customers=("Customer_ID", "nunique"),
        Unique_Dates=("Order_Date", "nunique"),
        Unique_Products=("Product", "nunique"),
        Total_Sales=("Total_Sales", "sum")
    )
    .query("Row_Count > 1")
    .reset_index()
)

display(order_id_summary)

print("Repeated ID groups:", len(order_id_summary))
```

```python

# DAY 5 - STEP 6: Sales validation

expected_sales = (
    df_day5["Quantity"] *
    df_day5["Unit_Price"]
).round(2)

recorded_sales = df_day5["Total_Sales"].round(2)

sales_mismatch_mask = (
    expected_sales.notna()
    & recorded_sales.notna()
    & expected_sales.ne(recorded_sales)
)

sales_mismatches = df_day5.loc[
    sales_mismatch_mask,
    [
        "Order_ID",
        "Quantity",
        "Unit_Price",
        "Total_Sales"
    ]
].copy()

sales_mismatches["Expected_Sales"] = (
    expected_sales.loc[sales_mismatch_mask]
)

print("Sales calculation mismatches:",
      len(sales_mismatches))

display(sales_mismatches.head(20))
```

```python

# DAY 5 - STEP 7: Business rule checks

business_checks = {
    "Age below 0 or above 120":
        ~df_day5["Age"].between(0, 120),

    "Quantity less than or equal to 0":
        df_day5["Quantity"] <= 0,

    "Unit_Price below 0":
        df_day5["Unit_Price"] < 0,

    "Total_Sales below 0":
        df_day5["Total_Sales"] < 0
}

business_report = []

for check_name, mask in business_checks.items():
    count = int(mask.fillna(False).sum())

    business_report.append({
        "Check": check_name,
        "Issue_Count": count,
        "Status": "Review required" if count else "Passed"
    })

business_report = pd.DataFrame(business_report)

display(business_report)
```

```python
# DAY 6 - STEP 1: Load verified Day 5 dataset

import pandas as pd
from pathlib import Path

day5_output = Path("../data/processed/day5_cleaned_dataset.csv")

df_day6 = pd.read_csv(
    day5_output,
    parse_dates=["Order_Date"]
)

print("Day 6 dataset loaded successfully!")
print("Shape:", df_day6.shape)

display(df_day6.head())
```

```python
# DAY 6 - STEP 2: Extract date features

df_day6["Order_Year"] = df_day6["Order_Date"].dt.year
df_day6["Order_Month"] = df_day6["Order_Date"].dt.month
df_day6["Order_Month_Name"] = df_day6["Order_Date"].dt.month_name()
df_day6["Order_Quarter"] = (
    "Q" + df_day6["Order_Date"].dt.quarter.astype(str)
)
df_day6["Order_Day"] = df_day6["Order_Date"].dt.day
df_day6["Order_Day_Name"] = df_day6["Order_Date"].dt.day_name()

print("Date features created successfully!")
display(
    df_day6[
        [
            "Order_Date",
            "Order_Year",
            "Order_Month",
            "Order_Month_Name",
            "Order_Quarter",
            "Order_Day",
            "Order_Day_Name"
        ]
    ].head(10)
)
```

```python
# DAY 6 - STEP 3: Monthly sales analysis

monthly_sales = (
    df_day6
    .groupby(
        ["Order_Year", "Order_Month", "Order_Month_Name"],
        as_index=False
    )
    .agg(
        Total_Orders=("Order_ID", "count"),
        Total_Quantity=("Quantity", "sum"),
        Total_Revenue=("Total_Sales", "sum"),
        Average_Order_Value=("Total_Sales", "mean")
    )
)

monthly_sales = monthly_sales.sort_values(
    ["Order_Year", "Order_Month"]
).reset_index(drop=True)

# Round monetary values for display
monthly_sales["Total_Revenue"] = monthly_sales["Total_Revenue"].round(2)
monthly_sales["Average_Order_Value"] = (
    monthly_sales["Average_Order_Value"].round(2)
)

print("Monthly sales analysis completed!")
display(monthly_sales)
```

```python
# DAY 9 - STEP 2: Missing-value audit

missing_report = pd.DataFrame({
    "Column": df_day9.columns,
    "Missing_Count": [
        df_day9[col].isna().sum()
        for col in df_day9.columns
    ],
    "Missing_Percentage": [
        round(df_day9[col].isna().mean() * 100, 2)
        for col in df_day9.columns
    ]
})

missing_report = missing_report.sort_values(
    "Missing_Count",
    ascending=False
)

display(missing_report)

print("Total missing cells:", df_day9.isna().sum().sum())
```

```python
# DAY 9 - STEP 3: Duplicate audit

exact_duplicates = int(df_day9.duplicated().sum())

repeated_order_rows = df_day9[
    df_day9.duplicated(subset=["Order_ID"], keep=False)
].sort_values("Order_ID")

print("Exact duplicate rows:", exact_duplicates)
print(
    "Rows with repeated Order_ID:",
    len(repeated_order_rows)
)
print(
    "Number of repeated Order_ID groups:",
    repeated_order_rows["Order_ID"].nunique()
)

display(repeated_order_rows)
```

```python
# DAY 9 - STEP 4: IQR outlier detection

numeric_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]

outlier_results = []

for col in numeric_columns:
    q1 = df_day9[col].quantile(0.25)
    q3 = df_day9[col].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (
        (df_day9[col] < lower_bound)
        | (df_day9[col] > upper_bound)
    )

    outlier_results.append({
        "Column": col,
        "Q1": round(q1, 2),
        "Q3": round(q3, 2),
        "IQR": round(iqr, 2),
        "Lower_Bound": round(lower_bound, 2),
        "Upper_Bound": round(upper_bound, 2),
        "Outlier_Count": int(outlier_mask.sum())
    })

outlier_report = pd.DataFrame(outlier_results)

display(outlier_report)
```

```python
# DAY 9 - STEP 5: Boxplots

for col in numeric_columns:
    plt.figure(figsize=(8, 4))
    plt.boxplot(df_day9[col].dropna(), vert=False)
    plt.title(f"Outlier Inspection: {col}")
    plt.xlabel(col)
    plt.tight_layout()
    plt.show()
```

```python
# DAY 9 - STEP 6: Sales calculation audit

expected_sales = (
    df_day9["Quantity"] * df_day9["Unit_Price"]
).round(2)

recorded_sales = df_day9["Total_Sales"].round(2)

sales_mismatch_mask = (
    expected_sales.notna()
    & recorded_sales.notna()
    & expected_sales.ne(recorded_sales)
)

sales_audit = df_day9.loc[
    sales_mismatch_mask,
    ["Order_ID", "Quantity", "Unit_Price", "Total_Sales"]
].copy()

sales_audit["Expected_Sales"] = expected_sales.loc[
    sales_mismatch_mask
]

print("Sales calculation mismatches:", len(sales_audit))
display(sales_audit)
```

```python
# DAY 9 - STEP 7: Business-rule validation

business_checks = {
    "Age outside 0-120": ~df_day9["Age"].between(0, 120),
    "Quantity <= 0": df_day9["Quantity"] <= 0,
    "Unit_Price < 0": df_day9["Unit_Price"] < 0,
    "Total_Sales < 0": df_day9["Total_Sales"] < 0
}

business_results = []

for check_name, mask in business_checks.items():
    issue_count = int(mask.fillna(False).sum())

    business_results.append({
        "Check": check_name,
        "Issue_Count": issue_count,
        "Status": "Review required" if issue_count else "Passed"
    })

business_report = pd.DataFrame(business_results)

display(business_report)
```

```python
# DAY 10 - STEP 1: Load final working dataset

import pandas as pd
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

project_dir = Path("..")
processed_dir = project_dir / "data" / "processed"
reports_dir = project_dir / "reports"

processed_dir.mkdir(parents=True, exist_ok=True)
reports_dir.mkdir(parents=True, exist_ok=True)

source_file = processed_dir / "day6_feature_engineered_dataset.csv"

df_final = pd.read_csv(
    source_file,
    parse_dates=["Order_Date"]
)

print("Final working dataset loaded!")
print("Rows:", df_final.shape[0])
print("Columns:", df_final.shape[1])

display(df_final.head())
```

```python
# DAY 10 - STEP 2: Final data-quality checks

expected_sales = (
    df_final["Quantity"] * df_final["Unit_Price"]
).round(2)

recorded_sales = df_final["Total_Sales"].round(2)

sales_mismatches = int(
    (
        expected_sales.notna()
        & recorded_sales.notna()
        & expected_sales.ne(recorded_sales)
    ).sum()
)

final_checks = pd.DataFrame({
    "Check": [
        "Expected row count",
        "Missing cells",
        "Exact duplicate rows",
        "Invalid order dates",
        "Sales calculation mismatches",
        "Invalid ages",
        "Non-positive quantities",
        "Negative unit prices",
        "Negative total sales"
    ],
    "Result": [
        len(df_final) == 1000,
        int(df_final.isna().sum().sum()),
        int(df_final.duplicated().sum()),
        int(df_final["Order_Date"].isna().sum()),
        sales_mismatches,
        int((~df_final["Age"].between(0, 120)).fillna(False).sum()),
        int((df_final["Quantity"] <= 0).fillna(False).sum()),
        int((df_final["Unit_Price"] < 0).fillna(False).sum()),
        int((df_final["Total_Sales"] < 0).fillna(False).sum())
    ]
})

display(final_checks)
```

```python
# DAY 10 - STEP 3: Final dataset summary

final_column_summary = pd.DataFrame({
    "Column": df_final.columns,
    "Data_Type": [
        str(df_final[col].dtype)
        for col in df_final.columns
    ],
    "Missing_Count": [
        int(df_final[col].isna().sum())
        for col in df_final.columns
    ],
    "Unique_Values": [
        int(df_final[col].nunique())
        for col in df_final.columns
    ]
})

display(final_column_summary)
```

```python
# DAY 10 - STEP 4: Final business KPIs

final_kpis = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Revenue",
        "Total Quantity Sold",
        "Average Sales per Record",
        "Unique Customers",
        "Unique Products",
        "Unique Categories",
        "Unique City Values"
    ],
    "Value": [
        len(df_final),
        round(df_final["Total_Sales"].sum(), 2),
        int(df_final["Quantity"].sum()),
        round(df_final["Total_Sales"].mean(), 2),
        int(df_final["Customer_ID"].nunique()),
        int(df_final["Product"].nunique()),
        int(df_final["Category"].nunique()),
        int(df_final["City"].nunique())
    ]
})

print("Final business KPIs:")
display(final_kpis)
```

```python
# DAY 10 - STEP 5: Monthly revenue report

monthly_summary = (
    df_final.groupby(
        ["Order_Year", "Order_Month", "Order_Month_Name"],
        as_index=False
    )
    .agg(
        Total_Records=("Order_ID", "count"),
        Total_Quantity=("Quantity", "sum"),
        Total_Revenue=("Total_Sales", "sum")
    )
    .sort_values(["Order_Year", "Order_Month"])
    .reset_index(drop=True)
)

monthly_summary["Total_Revenue"] = (
    monthly_summary["Total_Revenue"].round(2)
)

display(monthly_summary)

best_month = monthly_summary.loc[
    monthly_summary["Total_Revenue"].idxmax()
]

print("\nHighest-revenue month:")
print("Year:", best_month["Order_Year"])
print("Month:", best_month["Order_Month_Name"])
print("Revenue:", best_month["Total_Revenue"])
```

```python
# DAY 10 - STEP 6: Final category revenue chart

category_summary = (
    df_final.groupby("Category", as_index=False)
    .agg(Total_Revenue=("Total_Sales", "sum"))
    .sort_values("Total_Revenue", ascending=False)
)

plt.figure(figsize=(10, 6))

plt.bar(
    category_summary["Category"],
    category_summary["Total_Revenue"]
)

plt.title("Final Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()

display(category_summary)
```

```python
# DAY 10 - STEP 7: Save final dataset

final_dataset_path = (
    processed_dir / "day10_final_dataset.csv"
)

df_final.to_csv(
    final_dataset_path,
    index=False
)

print("Final dataset saved:")
print(final_dataset_path.resolve())
print("Rows:", len(df_final))
print("Columns:", len(df_final.columns))
```
